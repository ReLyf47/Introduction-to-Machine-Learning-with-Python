## 1. Scientific Python Libraries

* **NumPy**: Provides multi-dimensional arrays (`ndarray`) and vectorized mathematical routines.
* **SciPy**: Contains specialized functions for scientific computing, including memory-efficient sparse matrix storage algorithms (e.g., CSR, COO).
* **Pandas**: Offers tabular data handling via `DataFrame` and `Series` structures.
* **Matplotlib**: Primary library for static visualizations and diagnostic plots.

```python
# 1. Dense Matrix Representation (NumPy)
dense_arr = np.array([
    [1, 0, 0, 0],
    [0, 0, 2, 0],
    [0, 3, 0, 0]
])
print("Dense NumPy Array:\n", dense_arr)

# 2. Compressed Sparse Row (CSR) Format (SciPy)
# Compressed for row slicing and efficient matrix-vector multiplication
csr_matrix = sp.csr_matrix(dense_arr)
print("\nSciPy CSR Matrix:\n", csr_matrix)

# 3. Coordinate Format (COO) Sparse Matrix
# Constructed directly via non-zero values and coordinate tuples
values = np.array([4, 5])
row_indices = np.array([0, 2])
col_indices = np.array([1, 3])

coo_matrix = sp.coo_matrix((values, (row_indices, col_indices)), shape=(3, 4))
print("\nSciPy COO Matrix:\n", coo_matrix)
```

```python
# Tabular manipulation with Pandas
dataset_dict = {
    "Name": ["Alex", "Bianca", "Charlie", "Diana"],
    "Department": ["Computer Science", "Cybersecurity", "Data Science", "Software Eng"],
    "GPA": [3.85, 3.92, 3.50, 3.78],
    "CreditsCompleted": [90, 110, 45, 78]
}

df_students = pd.DataFrame(dataset_dict)
print("Student DataFrame:")
print(df_students)

# Filtering with boolean indexing
high_gpa = df_students[df_students["GPA"] >= 3.80]
print("\nHigh Performers (GPA >= 3.80):")
print(high_gpa)

# Functional plot using Matplotlib
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)
y = np.sin(x)

plt.figure(figsize=(8, 3.5))
plt.plot(x, y, color="teal", label=r"$y = \sin(x)$")
plt.title("Sample Sine Wave Plot")
plt.xlabel("Radians")
plt.ylabel("Amplitude")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

## 2. First Application: Classifying Iris Species

### Problem Formulation
* **Type**: Supervised Multiclass Classification
* **Goal**: Predict flower species based on continuous physical measurements.
* **Features ($X$)**: Sepal Length, Sepal Width, Petal Length, Petal Width (in centimeters).
* **Targets ($y$)**: `0` (Setosa), `1` (Versicolor), `2` (Virginica).

```python
iris_dataset = load_iris()

print("Dataset Keys:", iris_dataset.keys())
print("Target Names:", iris_dataset.target_names)
print("Feature Names:", iris_dataset.feature_names)
print("Data Matrix Shape:", iris_dataset.data.shape)
print("First 5 Samples:\n", iris_dataset.data[:5])
print("Target Output Shape:", iris_dataset.target.shape)
```

```python
iris_df = pd.DataFrame(iris_dataset.data, columns=iris_dataset.feature_names)

# Scatter Matrix colored by species target labels
pd.plotting.scatter_matrix(
    iris_df,
    c=iris_dataset.target,
    figsize=(9, 9),
    marker="o",
    hist_kwds={"bins": 20},
    s=40,
    alpha=0.8,
    cmap="viridis"
)
plt.suptitle("Pairwise Feature Scatter Matrix for Iris Dataset", y=0.92)
plt.show()
```

## 3. Model Training & Evaluation

### Train-Test Splitting
To evaluate generalization performance on unseen data, we split dataset into:
* **Training Set (75%)**: Used to build and train the estimator.
* **Test Set (25%)**: Held out to evaluate real-world model accuracy.

### $k$-Nearest Neighbors ($k$-NN)
Predictions are derived from the majority vote among the $k$ nearest training points using Euclidean distance:

$$d(\mathbf{a}, \mathbf{b}) = \sqrt{\sum_{i=1}^{n} (a_i - b_i)^2}$$

```python
# 1. Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset.data,
    iris_dataset.target,
    test_size=0.25,
    random_state=0
)

print(f"X_train shape: {X_train.shape} | y_train shape: {y_train.shape}")
print(f"X_test shape:  {X_test.shape}  | y_test shape:  {y_test.shape}")

# 2. Instantiate and Fit Estimator
knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train, y_train)

# 3. Predict on Unseen Sample
X_new = np.array([[5.0, 2.9, 1.0, 0.2]])
prediction_idx = knn.predict(X_new)

print("\n--- Model Inference ---")
print("Input Sample Features:", X_new)
print("Predicted Class Index:", prediction_idx[0])
print("Predicted Species Name:", iris_dataset.target_names[prediction_idx[0]])

# 4. Model Evaluation on Hold-Out Test Set
y_pred = knn.predict(X_test)
test_score = knn.score(X_test, y_test)

print("\n--- Model Performance ---")
print(f"Test Set Predictions: {y_pred}")
print(f"Test Set Accuracy:    {test_score * 100:.2f}%")
```
